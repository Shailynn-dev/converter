```python
choice = input("What will you be converting to? (Grams/Ml)?")
item = input("Please enter item:")
ml_converted = float(input("Enter the Ml to be converted: "))
grams_converted = float(input("Enter Grams to be converted: "))
```

```python

def item_calc():
    if choice == "grams":
        if item == "flour":
            grams = ml_converted * 0.50
            print(f"The amount in grams is: {grams}")
        elif item == "oats":
            grams = ml_converted * 0.34
            print(f"The amount in grams is: {grams}")
        elif item == "sugar":
            grams = ml_converted * 0.84
            print(f"The amount in grams is: {grams}")
        else:
            print("Invalid item.")
    else:
        if item == "flour":
            ml = grams_converted / 0.50
            print(f"The amount in mL is: {ml}")
        elif item == "oats":
            ml = grams_converted / 0.34
            print(f"The amount in mL is: {ml}")
        elif item == "sugar":
            ml = grams_converted / 0.84
            print(f"The amount in mL is: {ml}")
        else:
            print("Invalid item.")
```

```python
choice = input("Convert to grams or mL? ").lower()
item = input("Enter item (flour, oats, sugar): ").lower()

if choice == "grams":
    ml_converted = float(input("Enter the amount in mL: "))
else:
    grams_converted = float(input("Enter the amount in grams: "))

item_calc()
```
